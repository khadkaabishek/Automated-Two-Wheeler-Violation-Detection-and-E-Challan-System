```python
from ultralytics import YOLO
import os
BASE_DIR = '/content/drive/MyDrive/Two wheeler violation Detection models/Numberplate'
yaml_path = os.path.join(BASE_DIR, 'dataset.yaml')

# Initialize the YOLOv8 nano detection model weights
model = YOLO('yolov8n.pt')

# Train for 100 epochs using the free T4 cloud GPU
results = model.train(
    data=yaml_path,
    epochs=111,
    imgsz=640,
    batch=16,
    device=0,          # Set to 0 to target the T4 core accelerator
    workers=2,
    project=BASE_DIR,  # Directly dumps output curves/logs back to your drive folder
    name='yolov8_helmet_run',
    exist_ok=True
)
```
